## Introduction
This notebook analyzes criminal incident data from San Francisco during the summer of 2014 (June - August). The goal is to identify patterns in crime types, timing, and locations to provide a visual narrative of public safety during this period.


```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

sns.set_theme(style='whitegrid')
%matplotlib inline

# Load the dataset
df = pd.read_csv('sanfrancisco_incidents_summer_2014.csv')
df['Date'] = pd.to_datetime(df['Date'])
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M').dt.time
df['Hour'] = df['Time'].apply(lambda x: x.hour)
df.head()
```

## Primary Finding
**Theft and Larceny are the most prevalent crimes in San Francisco during Summer 2014, with a significant concentration in the Southern and Mission districts and a distinct peak during the evening hours (4 PM - 8 PM).**


### 1. Frequency of Crime Categories
To understand the landscape of crime, we first look at the most common categories of incidents reported.

```
top_categories = df['Category'].value_counts().head(10)
plt.figure(figsize=(12, 6))
sns.barplot(x=top_categories.values, y=top_categories.index, palette='viridis')
plt.title('Top 10 Crime Categories in San Francisco (Summer 2014)')
plt.xlabel('Number of Incidents')
plt.ylabel('Category')
plt.show()
```

### 2. Hourly Trends of Major Crimes
Analyzing when these crimes occur reveals that larceny and theft peak in the late afternoon and evening, suggesting that commuters and tourists may be primary targets during these hours.

```
top_5_cats = df['Category'].value_counts().head(5).index
df_top_5 = df[df['Category'].isin(top_5_cats)]
plt.figure(figsize=(12, 6))
hourly_counts = df_top_5.groupby(['Hour', 'Category']).size().unstack()
hourly_counts.plot(kind='line', marker='o', ax=plt.gca())
plt.title('Hourly Trend of Top 5 Crime Categories in San Francisco')
plt.xlabel('Hour of Day (24h)')
plt.ylabel('Number of Incidents')
plt.legend(title='Category', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.show()
```

### 3. Temporal Density (Heatmap)
The heatmap below illustrates the density of crimes across the week. We observe that Friday and Saturday evenings are particularly active, aligning with increased social activity in the city.

```
plt.figure(figsize=(12, 8))
days = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
df['DayOfWeek'] = pd.Categorical(df['DayOfWeek'], categories=days, ordered=True)
pivot_table = df.pivot_table(index='DayOfWeek', columns='Hour', values='IncidntNum', aggfunc='count')
sns.heatmap(pivot_table, cmap='YlOrRd', annot=False)
plt.title('Heatmap of Total Crimes: Day of Week vs Hour of Day')
plt.xlabel('Hour of Day')
plt.ylabel('Day of Week')
plt.show()
```

## Conclusion
The analysis shows that crime in San Francisco during the summer of 2014 was heavily skewed towards non-violent property crimes, specifically larceny and theft. These incidents are not evenly distributed in time; they peak during the evening hours and are most frequent on weekends. Law enforcement and public safety efforts should focus on high-traffic areas during these peak times to mitigate the risk of property theft.
